# Data Exploration

```python
import pandas as pd
from pathlib import Path
```

```python
data_dir = Path("/path/to/abeta-tiles")

csv_names = ["hold-out.csv", "train.csv", "validation.csv"]

dfs = []

for csv_name in csv_names:
    dfs.append(pd.read_csv(data_dir / csv_name))

df = pd.concat(dfs, ignore_index=True)

imagenames = df["imagename"]
filenames = [
    str(Path(n).name)
    for n in imagenames
]
len(df), len(set(filenames))
```
